```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "../data/processed/customer_segmentation_final.csv"

rfm = pd.read_csv(file_path)

segment_counts = (
    rfm["Segment"]
    .value_counts()
    .sort_values(ascending=False)
)

segment_counts
```

```python
segment_share = (
    rfm["Segment"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

segment_share
```
